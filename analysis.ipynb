# From a CSV to a saved result

Use this notebook for the data, files and Git tutorials.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

samples = pd.read_csv("data/samples.csv")
summary = samples.groupby("group")["signal"].agg(["mean", "count"])
print(summary)
Path("results").mkdir(exist_ok=True)
summary.to_csv("results/summary.csv")
summary["mean"].plot.bar(rot=0, color="#2686c9")
plt.ylabel("Mean signal")
plt.tight_layout()
plt.savefig("results/summary.png", dpi=160)
plt.show()
```
