## Read a small table
The measurements below are synthetic tutorial data.


```python
import pandas as pd
samples = pd.read_csv("data/samples.csv")
samples.head()
```

```python
import matplotlib.pyplot as plt
means = samples.groupby("group")["signal"].mean()
means.plot.bar(color=["#2686c9", "#48a990"], rot=0)
plt.ylabel("Mean signal (tutorial units)")
plt.title("A first plot in Vibeit")
plt.tight_layout()
plt.show()
```
